```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

X = df.drop(columns=['Level'])
from sklearn.preprocessing import LabelEncoder
y = df['Level']
le = LabelEncoder()
y = pd.Series(le.fit_transform(y))

suspicious_features = ['Obesity', 'Coughing of Blood', 'Alcohol use', 'Dust Allergy','Passive Smoker','Chest Pain','Wheezing']
print(f"Removing {len(suspicious_features)} suspicious features before training: {suspicious_features}\n")

X_safe = X.drop(columns=suspicious_features, errors='ignore')

X_train, X_test, y_train, y_test = train_test_split(X_safe, y, test_size=0.2, random_state=7)

rf = RandomForestClassifier(n_estimators=100, random_state=0)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
print("--- Tier 1 Model Performance (on safe features) ---")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', 
            xticklabels=['Low','Medium','High'], 
            yticklabels=['Low','Medium','High'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

prediction_metadata = {'label_encoder': le,'suspect_features': suspicious_features,'model_features': X_safe.columns.tolist()}
```

```python
from sklearn.feature_selection import mutual_info_classif
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("lung_cancer_data/cancer patient data sets.csv")  
df = df.drop(columns=['Patient Id','index'])

target_column = "Level"
X = df.drop(columns=[target_column])
y = df[target_column]

le = LabelEncoder()
y_numeric = pd.Series(le.fit_transform(y)) 

corr_with_target = X.corrwith(y_numeric)
corr_df = pd.DataFrame({"Feature": X.columns,"Correlation": corr_with_target,"AbsCorrelation": corr_with_target.abs()}).sort_values(by="AbsCorrelation", ascending=False)

mi_scores = mutual_info_classif(X, y_numeric, random_state=7)
mi_df = pd.DataFrame({"Feature": X.columns,"MutualInfo": mi_scores}).sort_values(by="MutualInfo", ascending=False)

corr_thresh = 0.7
mi_thresh = 0.5
suspect_features = set(corr_df[corr_df["AbsCorrelation"] > corr_thresh]["Feature"]) | \
                   set(mi_df[mi_df["MutualInfo"] > mi_thresh]["Feature"])

print("Suspicious features (possible leakage):", suspect_features)

X_train, X_test, y_train, y_test = train_test_split(X, y_numeric, test_size=0.2, random_state=7)
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
orig_acc = model.score(X_test, y_test)

X_dropped = X.drop(columns=suspect_features)
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(X_dropped, y_numeric, test_size=0.2, random_state=42)
model.fit(X_train_d, y_train_d)
drop_acc = model.score(X_test_d, y_test_d)

print(f"Accuracy (original): {orig_acc:.4f}")
print(f"Accuracy (dropped): {drop_acc:.4f}")
print(f"Accuracy drop: {orig_acc - drop_acc:.4f}")

y_shuffled = y.sample(frac=1, random_state=42).reset_index(drop=True)
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_dropped, y_shuffled, test_size=0.2, random_state=42)
model.fit(X_train_s, y_train_s)
shuffle_acc = model.score(X_test_s, y_test_s)
print(f"Accuracy after label shuffle: {shuffle_acc:.4f}")
```

```python
def symptom_present(value):
    if pd.isna(value):
        return False
    return value > 5

def safe_predict(patient_data, model, metadata):
    """
    Implements the two-tiered prediction logic:-
    Tier 2: Checks for override rules based on suspicious features.
    Tier 1: If no override, predicts using the model with safe features.
    """
   
    high_risk_symptoms = metadata['suspect_features']
    for feature in high_risk_symptoms:
        if feature in patient_data and symptom_present(patient_data[feature]):
            # Override triggered, predicts 'High' risk
            return {'prediction': 'High','reason': f"High-risk symptom '{feature}' was present with value {patient_data[feature]}."}

    # TIER 1: MODEL PREDICTION
    # If no override, prepares the data and uses the trained model
    patient_df = pd.DataFrame([patient_data])
    X_patient = patient_df[metadata['model_features']]

    prediction_numeric = model.predict(X_patient)[0]
    prediction_label = metadata['label_encoder'].inverse_transform([prediction_numeric])[0]

    return { 'prediction': prediction_label,'reason': 'No high-risk symptoms detected; prediction based on general features.'}
```
